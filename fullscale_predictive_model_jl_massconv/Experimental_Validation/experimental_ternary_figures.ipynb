```python
# Statistics for both fits
import numpy as np

def calc_fit_stats(exp, pred):
    # returns MAE, RMSE, and R² by component and overall
    exp = np.array(exp, dtype=float)
    pred = np.array(pred, dtype=float)
    residuals = pred - exp
    mae = np.mean(np.abs(residuals), axis=0)
    rmse = np.sqrt(np.mean(residuals**2, axis=0))
    r2_components = []
    for i in range(3):
        ss_res = np.sum((exp[:, i] - pred[:, i])**2)
        ss_tot = np.sum((exp[:, i] - np.mean(exp[:, i]))**2)
        r2_components.append(1 - ss_res / ss_tot)
    overall_rmse = np.sqrt(np.mean(residuals**2))
    overall_mae = np.mean(np.abs(residuals))
    overall_r2 = np.mean(r2_components)
    return mae, rmse, r2_components, overall_mae, overall_rmse, overall_r2

# Compute metrics
mae0, rmse0, r2c0, o_mae0, o_rmse0, o_r20 = calc_fit_stats(lit_organic, pred_organic_0)
mae25, rmse25, r2c25, o_mae25, o_rmse25, o_r225 = calc_fit_stats(lit_organic, pred_organic_2_5)

# Display results
print("="*60)
print("MODEL COMPARISON: τ = 0 vs τ = 2.5 (Organic Phase, Mole Fraction)")
print("="*60)
components = ["Water", "Furfural", "EA"]

for i, comp in enumerate(components):
    print(f"{comp:>8s} | τ=0 → MAE={mae0[i]:.4f}, RMSE={rmse0[i]:.4f}, R²={r2c0[i]:.4f}"
          f" | τ=2.5 → MAE={mae25[i]:.4f}, RMSE={rmse25[i]:.4f}, R²={r2c25[i]:.4f}")

print("-"*60)
print(f"Overall  | τ=0 → MAE={o_mae0:.4f}, RMSE={o_rmse0:.4f}, R²={o_r20:.4f}"
      f" | τ=2.5 → MAE={o_mae25:.4f}, RMSE={o_rmse25:.4f}, R²={o_r225:.4f}")
print("="*60)
```

Observation: Model performs better in the low-furfural region, which makes sense as that's what it was trained on.
Model performs better for EA/furfural predictions, which makes sense because the volume scales "mask" errors in predictions when in mole fraction space.
Aspen-trained tau zero data set very ideal -- the non-idealities from salt addition seem to somewhat compensate in the model space for the non-idealities from real solution behavior.
