```julia
using XLSX
using CSV
using DataFrames
using FilePathsBase

function excel_to_csv_all_sheets(excel_file::String, output_dir::String="")
    # Create output folder if not provided
    if isempty(output_dir)
        output_dir = "TrainingSets"
    end
    mkpath(output_dir)

    # Load the Excel file
    XLSX.openxlsx(excel_file) do xf
        for sheet in XLSX.sheetnames(xf)
            data = DataFrame(XLSX.readtable(excel_file, sheet))
            safe_name = replace(sheet, r"[^\w]+" => "_")
            csv_path = joinpath(output_dir, "$safe_name.csv")
            CSV.write(csv_path, data)
            println("Saved: $csv_path")
        end
    end
end


function main()
    excel_to_csv_all_sheets("Raw_Aspen_data.xlsx")
    println("Successully converted raw Excel data into CSV files. Please verify filepath is as intended.")
end

main()
```
